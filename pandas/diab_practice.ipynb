```python
def categorize_bmi(BMI):
    if BMI < 18:
        return 'underweight'
    elif 18 <= BMI <= 25:
        return 'fit'
    else:
        return 'overweight'
```

```python
df['bmi_category'] = df['BMI'].apply(categorize_bmi)
print("Data with new 'bmi_category' column:")
print(df[['BMI', 'bmi_category']].head())
```
