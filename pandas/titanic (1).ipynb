Name: The full name of the passenger.
Sex: The gender of the passenger (male or female).
Age: The age of the passenger.
SibSp: The number of siblings and/or spouses the passenger had aboard the Titanic.
Parch: The number of parents and/or children the passenger had aboard the Titanic.
Ticket: The ticket number of the passenger.
Fare: The fare the passenger paid for the ticket.
Cabin: The cabin number assigned to the passenger.
Embarked: The port where the passenger boarded the Titanic (C = Cherbourg, Q = Queenstown, S = Southampton).

#### The same axis keyword is used, but it has different implications depending on the operation

```python
def classdiv(class1):
    if class1 == 1:
        return 'business1'
    elif class1 == 2:
        return 'business2'
    else:
        return 'economic'

# Use apply with lambda to explicitly call the function
titanic['class2'] = titanic.apply(lambda x: classdiv(x['Pclass']), axis=1)

titanic
```

```python
def classdiv1(pclass, fare):
    if pclass == 1 and fare > 30:
        return 'luxury'
    elif pclass == 2 and fare > 20:
        return 'premium'
    elif pclass == 3:
        return 'economy'
    else:
        return 'basic'
```

```python
titanic['category1']=titanic[['Pclass','Fare']].apply(classdiv1)
titanic
```

```python
# Apply the function along columns considering both 'Pclass' and 'Fare'
titanic['category1'] = titanic.apply(lambda row: classdiv1(row['Pclass'],
                    row['Fare']),axis=1)#axis=1-->across the columns
titanic
```

```python
def classdiv1(pclass, fare):
    if pclass == 1 and fare > 30:
        return 'luxury'
    elif pclass == 2 and fare > 20:
        return 'premium'
    elif pclass == 3:
        return 'economy'
    else:
        return 'basic'
categories = []

for index, row in titanic.iterrows():
    category = classdiv1(row['Pclass'], row['Fare'])
    categories.append(category)

titanic['category1'] = categories

titanic
```
