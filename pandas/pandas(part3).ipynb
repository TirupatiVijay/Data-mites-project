```python
import pandas as pd
```

```python
weather=pd.read_csv('wea.csv')
weather
```

## Data Cleaning

```python
#nan-(missing value)
weather.isnull()
```

```python
weather.isnull().sum()
```

```python
weather.isnull().sum().sum()
```

```python
#1.filling the null values- fillna() methods inside are:
>by value()
>ffill - forward fill
>bfill-backward fill
>using dict
```

```python
weather.fillna(5)
```

```python
weather.fillna(method='ffill')
```

```python
#to provide condition such that it will not take multiple values in ffill (if first valus is null value we cannot fill with those value)
weather.fillna(method='ffill',limit=1) 
```

```python
weather.fillna(method='bfill')
```

```python
weather.fillna({'temp':23,'wind':8,'condition':'no event'})
```

```python
#2.replace
#df.replace(value to be replaced,with what value)
```

```python
import numpy as np
weather.replace(np.NaN,5)
```

```python
weather.replace([np.NaN,'Snow'],[85,'Snows'])
```

```python
#3.Drop method-dropna() stands for dropping the null values
weather.dropna()
```

```python
weather.dropna(how='any')
```

```python
weather.dropna(how='all') #it will delete a particular row if all the values in that row are null
```

```python
weather.dropna(thresh=2) #atleast 2 null values should be present in a row
```

```python
weather.dropna(thresh=4)
```
