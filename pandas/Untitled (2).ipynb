```python
def classdiv(class1):
    if class1==1:
        return 'business1'
    elif class1==2:
        return 'business2'
    else:
        return 'economis'
```

```python
titanic['class']=titanic['Pclass'].apply(classdiv)
```

```python
titanic
```

```python
def classdiv1(pclass,fare):
    if pclass==1 and fare>30:
        return 'luxury'
    elif pclass==2 and fare>20:
        return 'premium'
    elif pclass==3:
        return 'economy'
    else:
        return 'basic'
```

```python
titanic['category1']=titanic.apply(lambda row:classdiv1(row['Pclass'],row['Fare']),axis=1)
titanic
```
