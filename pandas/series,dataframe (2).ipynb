## Series

```python
#importing library
import pandas as pd
```

```python
#creating Series
```

```python
s1=pd.Series([1,2,3,4,5],name='num',dtype='float')
s1
```

```python
s1=pd.Series([1,2,3,4,5],index=('a','b','c','d','e'),
             name='num',dtype='float')
s1
```

```python
import numpy as np
import pandas as pd
s1=pd.Series(np.linspace(51,60,50),
             name='num',dtype='float')
s1
```

```python
res=pd.Series(arr1.reshape(15))
res
```

## DataFrame

```python
import pandas as pd
data=[['apple',1],['orange',4],['grapes',2],['mango'],
      ['kiwi',5]]
a=pd.DataFrame(data)
a
```

```python
a.columns=['fruits','quantity']
a
```

```python
a.index=['a','b','c','d','e']
a
```

```python
a
```

```python
 a.rename({'a':'ab'},axis=0,inplace=True)
 a
```

```python
d=pd.DataFrame(data,columns=['A','B'],index=['A','B','C','D','E']) 
## changing the index
d #keywords in dataframe are - columns,index,shape 
```

```python
detail={'name':['abi','poo','baru','shri'],
        'age':[23,22,23,22]}
det=pd.DataFrame(detail)
det#keys-->columns;values-->column values in dataframe 
```

```python
det.index=range(1,5)
det
```

```python
 det.rename({'age':'ages'},axis=1,inplace=True)
 det
```

```python
detail={'person1':{'name':'abi','age':23},
        'person2':{'name':'baru','age':22}}
det=pd.DataFrame(detail)
det#outer keys-->columns;inner keys-->indexes;inner values-->column values
```
