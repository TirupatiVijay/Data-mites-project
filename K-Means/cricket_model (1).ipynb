```python
from sklearn.preprocessing import MinMaxScaler
scale=MinMaxScaler()
df_scale=scale.fit_transform(df)
df_scale
```

```python
df2=pd.DataFrame(data=df_scale,columns=df.columns)
df2
```

```python
from sklearn.cluster import KMeans
wcss=[]
for i in range (1,12):
    #kmeans=KMeans(n_clusters=i,init='k-means++',max_iter=300,n_init=10,random_state=0)
    kmeans=KMeans(n_clusters=i,random_state=0)
    kmeans.fit(df2)
    wcss.append(kmeans.inertia_)
plt.plot(range(1,12),wcss,marker='o',linestyle='-',color='b') ## lineplot
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.xticks(np.linspace(0,11, 12, dtype=int))
plt.grid()
plt.show()
```

```python
from sklearn.cluster import KMeans
model = KMeans(n_clusters=3, random_state=42)

#Training the model
model.fit(df2) ## There is no training and testing in kneans as there is no target value so we fit the entire data
```

## K-MEANS METRIC

**Silhouette score:**

The calculation of silhouette score can be done using the following formula

         Silhouette score = (p-q) / max(p,q)

Here,

p --> mean distance to the points in the nearest cluster

q --> mean intra-cluster distance to all the points

It ranges from -1 to 1

- -1 --> wrong cluster

- 0 --> overlapping

- 1 --> best cluster

```python
from sklearn.metrics import silhouette_score
score=silhouette_score(df2,label)
score ## Above, 45 % is better
```
