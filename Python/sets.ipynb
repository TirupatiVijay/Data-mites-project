## Add and remove items

```python
myset={2,3,4} #Add item to a set using add() method     add() method is used to add single element to set
myset.add(9)
myset
```

```python
myset[1] #indexing is not supported in sets
```

```python
myset.update(['TEN','ELEVEN','TWELVE','TWELVE']) #update method is used to add multiple element to set
myset
```

```python
myset.remove(4) #remove method in a set using remove() method
myset
```

```python
myset.pop() #since unordered it will remove an element at any point in set
myset
```

```python
myset.discard(3) #remove an element from the set 
myset
```

```python
myset.clear()
myset
```

```python
del myset
myset
```

## set operations

```python
A={1,2,3,4,5}
B={4,5,6,7,8}
C={8,9,10} 
```

```python
A.union(B) #union of A and B #unique values is extracted
```

```python
B.union(A)
```

```python
A.union(B,C) #union of A,B and C
```

```python
A.intersection(B) #common values between 2 group values are extracted
```

```python
A.difference(B) #compare two set values and remove the common values and give priority to first set value and gets retrives
```

```python
B.difference(A)
```
