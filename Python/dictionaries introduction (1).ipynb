```python
"""
Dictionaries in Python are mutable, unordered collections of key-value pairs ***INDEXING IS DONE THROUGH KEYS NOT BY INDEX***.
They are useful for storing data that can be retrieved using unique keys.
Each key-value pair is separated by a colon (:), and pairs are separated by commas.
Dictionaries are defined with curly braces {}.
keys-->unique
values-->single,multiple,duplicate datas can be added
"""
```

```python
"""
Keys:
Analogy: The catalog uses unique identifiers for each book, such as ISBN numbers or library codes.
Python Dictionary: In a Python dictionary,
#keys are like these unique identifiers.
#used inorder to extract element values
They allow you to look up a specific item in the dictionary.
"""
```

```python
movie1={'title':'sachien','hero':'vijay','heroine':'jenelia'}
movie1
```

```python
#to extract title
movie1['title']
```

```python
#nested dictionaries
movie={'movie1':{'title':'sachin','hero':'vijay','heroin':'jenelia'},
       'movie2':{'title':'dragon','hero':'pradeep','heroin':'kayadu'}}
movie
```

```python
movie['movie1']['title']
```

```python
import pandas as pd
pd.DataFrame(movie) #to create a table form (combination of series)
```
