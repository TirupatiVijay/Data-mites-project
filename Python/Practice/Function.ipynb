```python
def sides():
    s1=int(input("enter side1:"))
    s2=int(input("enter side2:"))
    s3=int(input("enter side3:"))
    return s1,s2,s3
si1,si2,si3=sides()
print(si1,si2,si3)
if si1==si2==si3:
    print("eqi")
elif si1==si2 or si2==si3 or si3==si1:
    print("iso")
else:
    print("sca")
```

```python
def fact(num):
    prod=1
    for i in range(num,1,-1):
        prod=prod*i
    return prod
res=int(input("enter a number:"))
result=fact(res)
print("result is:",result)
```
