```python
def sum(*args):
    print(args)
    res=0
    for i in args:
        res=res+i
    return res
res1=sum(10,20,30,40)
res1
```

```python
def sum(name,*args):
    print(name,args)
    res=0
    for i in args:
        res=res+i
    return res
res1=sum('neha',10,20,30,40)
res1
```

```python
#factorial of a number
def factorial(num):
    prod=1
    for i in range(num,1,-1):
        prod=prod*i
    return prod
factorial(5)   
        
```

## Lambda functions

```python
#Lambda function
#In python are anonymous functions that can be defined without a name using the keyword lambda
#syntax- lambda arguments:expressions
#key points:
#No need for a def keyword:lambda functions are defined in one line without using 
#can take multiple arguments but should contain only one expression
#Expressions is evaluated and automatically returned without using return statement
```

```python
res=lambda a:a**2 #res=function name  a:a**2=function definition
res(5)
```

```python
#lambda with multiple arguments
res1=lambda a,b,c:a+b+c
res1(10,20,30)
```

```python
#reverse of a string
res2=lambda x:x[::-1]
res2('malayalam')
```

```python
res3=lambda str1,str2:str1+" "+str2
res3("hello","world")
```

```python
#with default arguments
res5=lambda x=3,y=2:x+y
print(res5(4,6))
print(res5())
```

# Filter

```python
#The filter() method filters the given sequence with the help of a function that tests each element in the sequence to be true or not.
#syntax-filter(function,iterable) #function can be of any type #iterables can be list,set,tuple,dict
#In backend process when applying filter it returns boolean values and then when typing print it will give the memory location so to view the result we need to forced typecasting
```

```python
#a list contains both even and odd numbers
seq=[0,1,2,3,5,8,13]
#result contains even numbers of the list
result=filter(lambda x:x%2==0,seq)
print(result)
list(result)
```

```python
seq=[0,1,2,3,5,8,13]
result=filter(lambda x:x%2!=0,seq)
print(result)
list(result)
```

```python
words=['apple','banana','cherry','avacoda','apricot']
res9=filter(lambda x:x.startswith('a'),words)
list(res9)
```

```python
res10=filter(lambda y:y.count('a')>2,words)
list(res10)
```

## Mapping

```python
#The map() function in python is used to apply a function to each item of an iterable(like set,tuple,list) and returns a map object(an iterator)
#syntax-map(function,iterable)
```

```python
numbers=['1','2','3','4','5']
res9=map(int,numbers)
print(res9)
res10=list(res9)
res10
```

```python
#round off to 2 decimal places
numbers=[5.90,3.145,90.23,89.568]
res11=map(lambda x:round(x,2),numbers)
list(res11)
```

```python
#Add two lists uding map and lambda
numbers1=[1,2,3]
numbers2=[4.4,5.7,6]
result=map(lambda x,y:x+y,numbers1,numbers2) #element wise operation
print(list(result))
```

```python
#map with user defined functions
def addition(n):
    return n+n
numbers=(1,2,3,4)
result=map(addition,numbers)
print(list(result))
```

```python
#with lambda function
def addition(n):
    return n+n
numbers=(1,2,3,4)
result=map(lambda n:n+n,numbers)
print(list(result))
```

## Zip function

```python
#The zip() function packing of elements from iterables will be done
#It is used to combine two or more iterables(like tuples or list) element wise inta a single iterable
#syntax-zip(iterable 1,iterable2)
```

```python
list1=[1,2,3]
list2=['a','b','c']
zipped=list(zip(list1,list2))
zipped2=dict(zip(list1,list2))
print(zipped)
print(zipped2)
```

```python
#zip function with list comprehension
list1=[1,2,3]
list2=[4,5,6]
sums=[x+y for x,y in zip(list1,list2)]
print(sums)
```

```python
#unzipping
zipped=[(1, 'a'), (2, 'b'), (3, 'c')]
list1,list2=zip(*zipped)
print(list1)
print(list2)
```

## Enumerate

```python
#the enumerate() function in python is used to add a counter to an iterable and returns an enumerate object
#This is useful when you need both the index and the value from a list or any other iterable (change index value)
```

```python
fruits=['apple','banana','cherry']
#use enumerate to get index and value
for i in enumerate(fruits):
    print(i)
```

```python
fruits=['apple','banana','cherry']
#use enumerate to get index and value
for i in enumerate(fruits,start=1): #change index value
    print(i)
```

```python
num=[3,90,21,89]
res8=[i**3 for i in num]
res8
```

```python
for i in enumerate(res8,start=1):
    print(i)
```

```python
list3=['sunday','monday','tuesday','wednesday','thrusday','friday','saturday']
dict(enumerate(list3,start=1))
```

```python
#doubling the elements and extracting elements greater than 10 (lambda,map,filter) 
seq = [0, 1, 2, 3, 5, 8, 13]
res1 = filter(lambda x: x > 10, map(lambda x: x*x, seq))
print(list(res1))
```

## Reduce function

```python
#The reduce(function,sequence) function is used to apply a particular function passed in its argument to all of the list elements mentioned in the sequence passed along.This function is defined in "functools" module.
#sequential operation=the sequential execution of operations one after another , consecutive operations
```

```python
from functools import reduce
lis=[1,3,5,6,2]
print(reduce(lambda a,b:a*b,lis))
```

```python
#using reduce to compute maximum element in list
print(reduce(lambda a,b:a if a>b else b,lis))
```
