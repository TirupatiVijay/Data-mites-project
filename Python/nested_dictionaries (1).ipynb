### DICTIONARIES WITH MULTIPLE VALUES

```python
#Create a dictionary to store the names of theme park visitors 
#and a list of their favorite rides.
#DICTIONARIES WITH MULTIPLE VALUES
```

```python
favorite_rides ={ "Emma": ["Roller Coaster", "Ferris Wheel"], 
                 "Liam": ("Haunted House", "Bumper Cars"), 
                 "Olivia": ["Water Slide", "Swing Ride"],
                 "Krissy":'Giant Wheel'}
```

```python
#access and print the list of favourite rides of emma
favorite_rides['Emma'][0]
```

```python
#Add a new favorite ride to the list of favourite rides of olivia
favorite_rides['Olivia'].append('titan')
favorite_rides
```

```python
favorite_rides['Emma'].insert(1,"giant wheel")
favorite_rides
```

```python
#update a favorite ride for liam
rides=('titanic',)
favorite_rides['Liam']=favorite_rides['Liam']+rides
favorite_rides
```

```python
favorite_rides['Olivia'].pop(1)
favorite_rides
```

```python
favorite_rides['Emma'][1:]
```

```python
#forced typecasting method to update value in tuple
list1=list(favorite_rides['Liam'])
list1.append('columbus')
print(list1)
```

```python
list1=tuple(list1)
list1
```

```python
favorite_rides['Liam']=list1
favorite_rides
```

```python
favorite_rides.pop('Krissy')
favorite_rides
```

```python
favorite_rides['Olivia'][1]='hauston'
favorite_rides
```

```python
"""
Basics of Nested Dictionaries
A nested dictionary is a dictionary within another dictionary. 
This structure is useful for storing complex, hierarchical data.
Deals in backend process of 2D datas(structured|Tabular 
datas in Pandas Package)
"""
```

```python
nested_dict = {
    "person1": {"name": "Alice", "age": 25},
    "person2": {"name": "Bob", "age": 30}
}#person1,person2-->outer keys
#name,age-->inner keys

nested_dict
```

```python
#to extract values from nested dictionaries
nested_dict['person1']
```

```python
nested_dict['person1']['age']
```

```python
nested_dict['person1']['dep']='ece'
nested_dict
```

```python
nested_dict.update({'person3':{'name':'neha','age':23}})
nested_dict
```

```python
del nested_dict['person1']['dep'] #nested_dict['person1'].pop('dep')
nested_dict
```

```python
import pandas as pd
pd.DataFrame(nested_dict)
```
