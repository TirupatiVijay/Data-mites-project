```python
weight = 45
height = 150
height1 = height / 100  # converting cm to meters
bmi = weight / (height1 ** 2)
print("bmi:", bmi)

if bmi < 18.5:
    print("underweight")
elif 18.5 <= bmi < 25:
    print("normal weight")
else:
    print("overweight")
```
