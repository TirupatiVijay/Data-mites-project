## Task-2

```python
# Dictionary with employee names and their monthly salaries
salaries = {
    'John': [3000, 3200, 3100],
    'Emily': [3500, 3600, 3700],
    'Michael': [4000, 4200, 4100],
    'Sophia': [2800, 2900, 3000]
}
```

```python
Calculate the average salary for each employee over the given 3 months
and update the salaries dictionary with this sum as value
change should be in original dictionary
```

```python
res1=salaries['John']
res1
```

```python
res2=sum(res1)/3
res2
```

```python
salaries.update({'John':res3})
salaries
```

```python
res4=salaries['Emily']
res5=sum(res4)/3
res7=salaries['Michael']
res8=sum(res7)/3
res10=salaries['Sophia']
res11=sum(res10)/3
salaries.update({'Emily':res5,'Michael':res8,'Sophia':res11})
salaries
```
