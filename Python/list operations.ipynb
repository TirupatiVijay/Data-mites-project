```python
mixed_list =[90,78,56,True,'data',False,'datamites']
new_list=sorted(mixed_list,key=str) #str is used to convert all data types to string format
print(new_list) #numbers,uppercase,lowercase
```

```python
names=['baru','divya','neha','thiaga','shri','bars']
res=sorted(names,key=len) #replicate in same order of the elements provided in list 
print(res)
```
