```python
#given year is leap year or not - year%4 and not year%100 or year%400
```

```python
leap_year=int(input("enter a year:"))
if ((leap_year%4==0 and leap_year%100!=0) or (leap_year%400==0)):
    print("its a leap year")
```

```python
#using formating print statement
leap_year=int(input("enter a year:"))
if ((leap_year%4==0 and leap_year%100!=0) or (leap_year%400==0)):
    print(f'{leap_year} is leap year')
else:
    print(f'{leap_year} is not a leap year')
```

```python
#if-elif-else 3-conditions
temp=int(input("enter the temperature:"))
if temp>30:
    print("hot")
elif 15<=temp<=30: #temp>=15 and temp<=30
    print("moderate")
else:
    print("cold")
```

```python
weight=int(input("enter weight:"))
height=int(input("enter height:"))
height1=height/100
bmi=weight/(height1)**2
if bmi<18:
    print("underweight")
elif bmi>18 and bmi<25: 
    print("fit")
else:
    print("overweight")
```

```python
temp=int(input("enter temp:"))
if temp>0 and temp>30:
    print("extreme hot")
elif temp>0 and temp<=30:
    print("hot")
elif temp<0 and temp>=-15:
    print("cold")
else:
    print("extreme cold")
    
```

```python
#nested if -if within if block
temp=int(input("enter a temp:"))
if temp>0:
    if temp>30:
        print("extreme hot")
    else:
        print("hot")
else:
    if temp>=-15:
        print("cold")
    else:
        print("extreme cold")
```
