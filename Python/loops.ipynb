## FOR LOOP

```python
#for loop
#it is used to performe same and repeated task multiple times
#range is a built in function that creates a sequence of numbers
#syntax:range(start,stop,step) start is included,stop is excluded
#default: start-0,stop-1
# for loop purpose:
#Automatic and repeated tasks
#iterating over sequences
#performance operation on each and every element
#disadvantage-when range value is unknown we cannot proceed with for loop
```

```python
print(range(5))
```

```python
#to view the values in range we need to do forced tycasting 
list(range(5))
```

```python
list(range(50,61,2))
```

```python
list(range(100,89,-1)) #stop-1 -->downward stop+1 -->forward 
```

```python
for i in range(12):
    print(f'added choco chip for cookie {i+1}')
```

```python
for i in range(71,91,2): #to print odd numbers
    print(i)
```

```python
for i in range(169,149,-2): #reverse way of printing odd numbers
    print(i)
```

```python
#iterating of for loop
name='diya'
for i in name: #the loop will run 4 times as i take each and every character from name 'diya'
    print(i)
```

```python
details=['datamites',24,9.8]
for i in details: #i will take the entire elements in list
    print(i)
```

```python
details={'diya':15,'mites':30}
for i in details.keys():#to extract keys from dictionary
    print(i)
```

```python
details={'diya':15,'mites':30}
print(details.items()) #to extract items
for i in details.keys():
    print(i)
```

```python
details={'diya':15,'mites':30}
print(details.items()) #to extract items and bring individual values
for x,y in details.items():
    print(x,y)
```

```python
list1=['data','24','mites']
for i in list1:
    res=i[0] #to extract 1 character from each element
    print(res)
```

```python
#write a program that counts the number of even and odd numbers
numbers=[10,-21,32,-43,-54,65,-76,87,-98,-109]
even=[]
odd=[]
for i in numbers:
    if i%2==0:
        even.append(i)
    else:
        odd.append(i)
print(even,odd)
```

```python
words=['apple','banana','avocado','grape','apricot','berry']
#extract fruits that start with letter 'b'- startswith('b')
#extract the fruit that has the longest length 
#1.create a empty list length 
#2.use for loop in order to store length of each and every fruit
#3.use another for loop and check whether length of the fruit is equal to maximum length
#4.if codition satisfies print the corresponding fruit
```

```python
words = ['apple', 'banana', 'avocado', 'grape', 'apricot', 'berry']
words1 = []
for i in words:
    if i.startswith('b'):
        words1.append(i)
print(words1)
        
```

```python
length=[]
for i in words:
    length.append(len(i))
print(length)
    
```

```python
for i in words:
    if len(i)==7:
        print(i)
```

```python
max1=max(length)
for i in words:
    if len(i)==max1:
        print(i)
```

```python
#updation of variables
num=[10,-34,90,-67,-45,91]
c_p=0
c_n=0
for i in num:
    if i>0:
        c_p=c_p+1
    else:
        c_n=c_n+1
print(c_p,c_n)
```

```python
#sum of digits in a number
num=3571
sum=0
for i in range(4):
    rem=num%10
    num=num//10
    sum=sum+rem
print(sum)
```

```python
#reverse of the number
num=3571
sum=0
for i in range(4):
    rem=num%10
    num=num//10
    sum=sum*10+rem
print(sum)
```

```python
#Armstrong Number
num=153
sum=0
num1=num
for i in range(3):
    rem=num%10
    num=num//10
    sum=sum+rem**3
print(sum)
if num1==sum:
    print("armstrong number")
else:
    print("not armstrong")
```

```python
#nested for loop
fruits=['apple','banana','cherry']
for i in fruits: #outer for loop assigned
    for j in i: #inner for loop executes till the condition  then outer for loop gets updated
        print(j) 
```

```python
#pattern printing
#i-rows
#j-columns
for i in range(1,6):
    for j in range(1,6):
        print('*',end=" ") #end is used to maintain stars in same line
    print() #next line move
```

```python
#to print right angle triangle
for i in range(1,6):
    for j in range(1,i+1):
        print('*',end=" ")
    print()
```

```python
for i in range(1,6):
    for j in range(1,i+1):
        print(i,end=" ")
    print()
```

```python
for i in range(1,6):
    for j in range(1,i+1):
        print(j,end=" ")
    print()
```

```python
num=1
for i in range(1,6):
    for j in range(1,i+1):
        print(num,end=" ")
        num=num+1
    print()
```

```python
#reverse right angle triangle
for i in range(6,1,-1):
    for j in range(1,i):
        print('*',end=" ")
    print()
```

```python
num=5
for i in range(1,6):
    for j in range(1,num+1):
        print('*',end =" ")
    num=num-1
    print() 
```

```python
num=65
for i in range(1,6):
    for j in range(1,i+1):
        print(chr(num),end=" ")
    num=num+1
    print()
```

## While loop

```python
#while loop-->looping executes until condition satisfies
#it can support both known and unknown conditions
```

```python
i=5 #initialization
while i>0: #condition
    print(i)
    i=i-1 #increment or decrement
```

```python
i=50
while i<61:
    print(i)
    i=i+2
```

```python
i=50
sum=0
while i<71:
    sum=sum+i
    i=i+1
print(sum)
```

```python
#reverse of a num
num=int(input("enter a number"))
sum=0
num1=num
while num>0:
    rem=num%10
    num=num//10
    sum=sum*10+rem
print(sum)
if num1==sum:
    print("Palindrome")
else:
    print("not palindrome")
```
