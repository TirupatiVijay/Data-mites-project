```python
#How do you append an element to a list?
"""
append:

The append method takes a single element as an argument and adds it to the end of the list.
If the argument is a list, the entire list is added as a single element (a nested list).
extend:

The extend method takes an iterable as an argument and adds each element of the iterable to the end of the list.
This means that if the argument is a list, each element of that list is added to the original list, one by one.
"""
list1=[]
list1.append("electronics")
print(list1)
```

```python
list1.extend(['IT','CSE','DS','AI'])
print(list1)
```
