```python
import numpy as np
import pandas as pd
heart=pd.read_csv('heart.csv')
heart
```

### Null Hypothesis (H0): The average cholesterol level is less than or equal to 200 mg/dl.
### Alternative Hypothesis (H1): The average cholesterol level  is greater than 200 mg/dl.

```python
import scipy.stats as stats
cholesterol_levels = heart['chol']

# Set the sample size
sample_size = 100

# Randomly sample 100 cholesterol levels
sample = cholesterol_levels.sample(n=sample_size, random_state=42)

# Define the population mean (hypothesized mean)
population_mean = 200
```

```python
# Calculate statistics
sample_mean = sample.mean()#  # Mean of the sample
sample_std = sample.std(ddof=1)  # Sample standard deviation

# Z-test calculation
z_score = (sample_mean - population_mean) / (sample_std / (sample_size ** 0.5))
print(z_score)
```

```python
stats.norm.ppf(0.95)#critical value for one tailed test
```

```python
# Critical value for one-tailed test
critical_value = stats.norm.ppf(0.95)
# percent point function (PPF) gives the value below which a certain percentage 
#of data falls
results = {
    "Sample Mean": sample_mean,
    "Z-Score": z_score,
    "Critical Value at C.L=95%": critical_value
}

print(results)
```

```python
if z_score > critical_value:
    print("Accept Ha")
else:
    print("accept H0")
```

```python
#Z TEST USING P-value
#CDF stands for Cumulative Distribution Function.For a normal distribution, 
# the CDF tells you the area under the curve to the left of a given z-score.
C.L=0.95
alpha = 1-C.L#0.05    

p_value = 1 - stats.norm.cdf(z_score)

if p_value < alpha:
    print(f"Accept alternate hypothesis (P-value: {p_value})")
else:
    print(f"Accept null hypothesis (P-value: {p_value})")

```

```python
#T-test
#we can compare the mean cholesterol (chol) levels between two groups:
#Group 1: Patients with heart disease (target = 1)
#Group 2: Patients without heart disease (target = 0)
```

#### Null Hypothesis (H₀): There is no difference in the mean cholesterol levels between the two groups.
#### Alternative Hypothesis (H₁): There is a significant difference in the mean cholesterol levels between the two groups.

```python
sample_size = 25  # n < 30

group1 = heart[heart['target'] == 1]['chol'].sample(n=sample_size, random_state=42)
group2 = heart[heart['target'] == 0]['chol'].sample(n=sample_size, random_state=42)
```

```python
# Perform an independent t-test
t_stat, p_value = stats.ttest_ind(group1, group2)

alpha = 0.05

print(f"T-statistic: {t_stat}")
print(f"P-value: {p_value}")
```

```python
heart.corr()
```

```python
if p_value < alpha:
    print("accept alternate hypothesis: There is a significant difference in mean cholesterol levels.")
else:
    print("accept null hypothesis: No significant difference in mean cholesterol levels.")
```

##### Hypotheses:
##### Null Hypothesis (H₀): treatment had no effect on patients
##### Alternative Hypothesis (H₁): treatment has made changes in patients

```python
before = np.array([100, 102, 98, 105, 110, 99, 101, 103, 97, 100])
after = np.array([102, 100, 99, 107, 108, 98, 102, 101, 96, 101])
```

```python
# Perform paired t-test
t_stat, p_value = stats.ttest_rel(before, after)

# Print the results
print(f"T-statistic: {t_stat}")
print(f"P-value: {p_value}")

# Set the significance level
alpha = 0.05
```

```python
# Hypothesis test conclusion based on P-value alone
if p_value < alpha:
    print("accept alternate hypothesis: There is a significant difference between the paired samples.")
else:
    print("accept null hypothesis: No significant difference between the paired samples.")
```

```python
# Sample size for each 'cp' type (ensure n < 30)
sample_size = 10  # Adjust as needed
samples = []

# Get unique 'cp' values
cp_values = heart['cp'].unique()

for cp in cp_values:
    # Randomly sample from the 'chol' column for each 'cp' type
    sample = heart[heart['cp'] == cp]['chol'].sample(n=sample_size, random_state=42)
    samples.append(sample)
print(samples)
```

```python
f_statistic, p_value = stats.f_oneway(*samples)

# Print the results
print(f"F-statistic: {f_statistic}")
print(f"P-value: {p_value}")

# Set the significance level
alpha = 0.05
```

```python
# Hypothesis test conclusion based on P-value
if p_value < alpha:
    print("accept alternate hypothesis: There is a significant difference in mean cholesterol levels among the chest pain types.")
else:
    print("accept null hypothesis: No significant difference in mean cholesterol levels among the chest pain types.")
#cholestrol level and cp column are uncorrelated
```
