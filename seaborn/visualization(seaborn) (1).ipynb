```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
df=sns.load_dataset("iris")
df
```

```python
df.species.unique()
```

```python
df.species.value_counts()
```

```python
df.describe()
```

```python
#line plot 
#univariate-visualization with respect to single column
#bivariate-visaulization with respect to two columns
#multivariate-more than 2 columns
```

```python
sns.lineplot(data=df,x=df.index,y='sepal_length',markers='*',color='green')
```

```python
sns.lineplot(data=df,x=df.index,y='sepal_length',markers='*',color='green')
plt.show()
```

```python
import warnings
warnings.filterwarnings('ignore')
sns.lineplot(data=df,x='sepal_length',y='sepal_width',markers='*')
plt.show()
```

```python
import warnings
warnings.filterwarnings('ignore')
sns.lineplot(data=df,x='sepal_length',y='sepal_width',markers='*',ci=None)
plt.show() #ci-->confidence interval(futuristic values of features will be provided ci=95%)
```

```python
sns.lineplot(x='sepal_length',y='sepal_width',data=df,hue='species',ci=None)
plt.show() #hue-define colour varaitions depending on the categories for a specified column
```

```python
df.groupby('species')['sepal_length'].agg(['min','max'])
```

```python
#continuous and categorial feature
sns.lineplot(x='sepal_length',y='species',hue='species',data=df,ci=None)
plt.show()
```

```python
##scatter plot to understand the correlation between the species
sns.scatterplot(x='petal_length',y='petal_width',data=df,hue='species')
plt.title('Scatter plot')
plt.show()
```

```python
sns.scatterplot(x='petal_length',y='sepal_length',data=df,hue='species')
plt.title('Scatter plot')
plt.show()
```

```python
#bar plot-categorical estimate plot()
#creates a plot taking a mean per categorical column
sns.barplot(x='species',y='sepal_length',data=df,hue='species')
plt.show()
```

```python
df.groupby('species')['sepal_length'].agg(['mean'])
```

```python
df.groupby('species')['sepal_length'].mean()
```

```python
#count plot-->returns value_counts()
##A countplot basically counts the categories and returns a count of their occurances.
sns.countplot(x='sepal_length',data=df,palette='rainbow')
plt.show()
```

```python
#to avoid overlap of values we include figure to increase the size 
plt.figure(figsize=(15,10))
sns.countplot(x='sepal_length',data=df,palette='rainbow')
plt.show()
```

```python
df['sepal_length'].value_counts()
```

```python
##box plot also called whisker plot, presence of outliers
sns.boxplot(x='sepal_width',data=df)
plt.show()
```

```python
sns.boxplot(x='species',y='sepal_width',data=df)
plt.show() #checking outliers in each category with respect to sepal width
```

```python
sns.boxplot(data=df)
plt.show()
```

```python
#violinplot-->boxplot+density information
sns.violinplot(x='species',y='sepal_width',data=df)
plt.show()
```

```python
##stripplot
#It is basically creates a scatter plot based on the category
#it is created using the stripplot()
sns.stripplot(x='species',y='sepal_width',data=df,hue='species')
plt.show()
```

```python
#swarmplot-scatterplot+countplot
sns.swarmplot(data=df,x='species',y='sepal_length',hue='species')
```

```python
##Histogram-is basically used to represent data provided in a form of some groups
sns.histplot(x='sepal_width',data=df,bins=10,kde=True)
plt.show()
#kde=kernel density estimator
```

```python
sns.histplot(x='sepal_width',data=df,bins=10,hue='species',kde=True)
plt.show()
```

```python
##pairplot
##pairplot represents paiwise relation across the enitre dataframe
#Multivariant Analysis,hue is used for categorical separation
sns.pairplot(data=df,hue='species')
plt.show() 
```

```python
df[df.columns[:-1]].corr()
```

```python
##Heat map
#Heat map is defined as a graphical representation of data using colors
tc=df[df.columns[:-1]].corr()
tc
```

```python
sns.heatmap(tc)
plt.show()
```

```python
#to view values annot is used -Annotation
sns.heatmap(tc,annot=True)
plt.show() #fmt(format)-'g','d' -->extact integer value will be returned
```
