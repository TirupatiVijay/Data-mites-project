```python
import numpy as np

python_scores = np.array([10, 20, 15, 20, 12, 15, 8, 5, 13, 17])

# Calculate unique values and their counts
unique_values, counts = np.unique(python_scores, return_counts=True)
print(unique_values)
# Find the maximum count(s)
max_count = np.max(counts)

# Find the indices where counts are equal to the maximum count
mode_indices = np.where(counts == max_count)
print(mode_indices)
print(mode_indices[0])
# The modes are the unique values at these indices
modes = unique_values[mode_indices[0]]

print("Modes:", modes)
```
