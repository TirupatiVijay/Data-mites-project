### Fancy Indexing

Fancy indexing allows you to select entire rows or columns out of order,to show this, let's quickly build out a numpy array:

```python
#Set up matrix
arr2 = np.zeros((10,10),dtype='int')  # taking shape from an existing array
```

```python
arr2
```

```python
arr2.shape[0]
```

```python
rows,columns=arr2.shape
print(rows)
```

```python
no_of_rows=arr2.shape[0]  # indexing the shape tuple
```

```python
#Set up array

for i in range(no_of_rows):
    arr2[i] = i+1

arr2
```

```python
import numpy as np
arr2=np.random.randint(0,1001,100).reshape(10,10)
arr2
```

Fancy indexing allows the following

```python
arr2[[6,3,8]]#for rows
```

```python
arr2[:,[2,4,6,8]]#for columns
```

```python
#Allows in any order
arr2[[6,4,5,7],[3,4,5,7]]#doesnt extract specific rows and columns
```

```python
arr1=np.array(range(1,17))
arr1
```

```python
arr2=arr1.reshape(4,4)
arr2
```

```python
arr2[[0,1,2,3],[0,1,2,3]]
```
