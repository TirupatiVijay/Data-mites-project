### arange(start,stop,step)

Return evenly spaced values within a given interval.

-->range() generates a sequence of integers.
-->np.arange() generates a NumPy array, which can contain integers, floating-point numbers, or other data types.

-->range() only supports integer step sizes.
-->np.arange() allows floating-point step sizes.

```python
np.arange(15) # end; default start at 0
```

```python
np.arange(50,101,2.5).reshape(3,7) # start, end and step
```

```python
np.arange(0,10,2) # start end and step
```

```python
np.arange(1,11,2.5)
```

```python
np.arange(0,10,1.523)
```

```python
import numpy as np
np.round(np.arange(0,10,1.523),2)
```

### Broadcasting -Broadcasting is a mechanism that allows NumPy to perform element-wise operations on arrays of different shapes

```python
import numpy as np
arr1 = np.array([4,9,16])
arr1+10
```

```python
arr2=np.array([5,9,10])
arr1+arr2
```

```python
arr1 = np.array([4,9,16])## scalar value used for each element of the array
x=arr1
print('x+5:', x+5) #addition
print('x-5:', x-5) #subtraction
print('x*2:', x*2) #multiplication
print('x/2:', x/2) #division
print('x//2:', x//2) #floor division - Greatest Integer Function
print('x**2:', x**2) #Power
print('x%2:', x%2) #remainder
```

```python
arr2=np.array([[1,2,3],[4,5,6]])
arr2
```

```python
arr2+[1,3,5]
```

```python
arr1+[1,5,3]
```

### max,min,argmax,argmin

numpy.argmin(): Returns the index of the minimum value in an array.

numpy.argmax(): Returns the index of the maximum value in an array.

numpy.unique(): Returns the unique elements of an array.

```python
import numpy as np
arr4=np.arange(10,50,7)
arr4
```

```python
np.argmax(arr4)
```

```python
np.argmin(arr4)
```
